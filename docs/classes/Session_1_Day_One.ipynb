## <span class="hz-step-n">02</span> Variables and types

Let's describe one customer, CM-10482, the way the billing system does: a plan, a bill, how long they have been with us, and whether they pay automatically. Each fact gets a **variable**: a name on the left of `=`, a value on the right.

A variable is a **label**, not a box: the value lives in memory and the name points at it. When you write `tenure_months = tenure_months + 1`, Python first computes the right side, then moves the label to the new value. Step through it.

```python
customer_id = "CM-10482"
plan = "Plus"
monthly_bill = 31.9
tenure_months = 14
autopay = True

print(customer_id, plan, monthly_bill, tenure_months, autopay)
```

Every value has a **type**, and the type decides what you can do with it. You will meet four all the time:

| type | what it holds | Cordillera example |
|---|---|---|
| `str` | text, in quotes | `"Plus"`, `"CM-10482"` |
| `int` | whole numbers | `14` months |
| `float` | numbers with decimals | `31.9` USD |
| `bool` | `True` or `False` | `autopay = True` |

```python
print(type(plan))
print(type(tenure_months))
print(type(monthly_bill))
print(type(autopay))
```

Data often arrives as text: a CSV export or a web form gives you `"14"`, not `14`. Text and numbers don't mix, so you **convert** with `int()`, `float()` or `str()`.

To write sentences for a report, use an **f-string**: put an `f` before the quotes and any variable inside `{ }`. Add `:.2f` to show two decimals, or `:,.2f` to also add a thousands separator.

```python
tenure_text = "14"          # as it came from a web form

try:
    tenure_text + 1
except Exception as e:
    print(type(e).__name__, "→", e)

print(int(tenure_text) + 1)     # convert first, then add
print(float("31.9") * 2)
print("Months: " + str(tenure_months))
```

```python
# f-strings: variables inside { }, formats after a colon
print(f"Customer {customer_id} is on the {plan} plan and has been with us {tenure_months} months.")
print(f"Last bill: {monthly_bill:.2f} USD. A year at this price: {monthly_bill * 12:,.2f} USD.")
```

## <span class="hz-step-n">03</span> Lists: a customer's last six bills

One bill says little. The commercial team wants to see the **trend**: the last six monthly bills of CM-10482, April to September. A **list** keeps many values in order, inside square brackets.

Each item has a position, its **index**, and Python counts from **0**. Negative indices count from the end, so `bills[-1]` is always the latest bill, however long the list is. A **slice** `bills[start:stop]` takes a range and never includes `stop`. Try some expressions.

```python
months = ["Apr", "May", "Jun", "Jul", "Aug", "Sep"]
bills = [29.4, 30.8, 30.1, 35.6, 32.3, 31.9]

print("First month:", months[0], bills[0])
print("Latest month:", months[-1], bills[-1])
print("Last three bills:", bills[-3:])

plans = ["Prepaid", "Basic", "Plus", "Unlimited"]     # lists can hold text too
print(len(plans), "plans; is 'Plus' one of them?", "Plus" in plans)
```

Lists come with tools that answer business questions in one line: how many items (`len`), the total (`sum`), the extremes (`min`, `max`).

```python
average = sum(bills) / len(bills)

print(f"{len(bills)} bills, {sum(bills):.2f} USD in total")
print(f"Average bill: {average:.2f} USD (lowest {min(bills)}, highest {max(bills)})")
print("Highest month:", months[bills.index(max(bills))])
```

The July spike (35.6 USD) is the kind of "bill shock" the retention team worries about. Lists can also grow: `append` adds an item at the end. October's bill just came in. We rebuild the list inside the cell on purpose: with a plain `bills.append(33.0)`, every extra run of the cell would add October again (the "cell run twice" trap from the first explainer).

```python
october = [29.4, 30.8, 30.1, 35.6, 32.3, 31.9]   # a fresh copy of the six bills
october.append(33.0)                              # add October's bill
print(october, "→", len(october), "bills")
```

## <span class="hz-step-n">04</span> Dictionaries: one customer, one record

Five loose variables for one customer gets messy fast. A **dictionary** (`dict`) keeps all the facts about the customer together, each under a name called a **key**: `customer["plan"]` reads like a column name in a spreadsheet row.

You look a value up by its key, not by position. Asking for a key that isn't there raises a `KeyError`; `.get(key, default)` returns a default instead. Try both.

## <span class="hz-step-n">05</span> Conditionals: the retention team's first rule

The retention team has a rule of thumb, written on a whiteboard:

- **at risk** if the customer is on a **month-to-month** contract (free to leave any day) **and** has **3 or more** support tickets;
- **watch** if **only one** of the two is true;
- **ok** otherwise.

In Python that is `if` / `elif` / `else`. Python checks the conditions from the top and runs **only the first branch that is true**; the rest are skipped. Move the controls and watch the path.

Conditions are built from **comparisons** (`==` equal, `!=` different, `>`, `>=`, `<`, `<=`), which give `True` or `False`, combined with `and` (both must be true), `or` (at least one) and `not`.

```python
contract = customer["contract"]
support_tickets = customer["support_tickets"]

print(contract == "Month-to-month")
print(support_tickets >= 3)
print(contract == "Month-to-month" and support_tickets >= 3)
```

```python
if contract == "Month-to-month" and support_tickets >= 3:
    risk = "at risk"
elif contract == "Month-to-month" or support_tickets >= 3:
    risk = "watch"
else:
    risk = "ok"

print(customer["customer_id"], "→", risk)
```

CM-10482 is month-to-month but has "only" 2 tickets, so they land on "watch": one more complaint and they would be at risk. Notice the **indentation**: the 4 spaces tell Python which lines belong to each branch. Now a different customer, on a 12-month contract with one ticket:

```python
contract = "12 months"
support_tickets = 1

if contract == "Month-to-month" and support_tickets >= 3:
    risk = "at risk"
elif contract == "Month-to-month" or support_tickets >= 3:
    risk = "watch"
else:
    risk = "ok"

print(risk)
```

## <span class="hz-step-n">06</span> Loops: every customer, one rule

We just copied the whole rule to check a second customer; with five thousand customers, copying is not an option. Two numbers are due at Monday's meeting: the commercial team wants the **total monthly revenue** of the watch list, and the retention team wants **how many customers are at risk**.

A `for` loop runs the same lines once for **each** item of a list. The loop variable (`c` here) takes each customer in turn, and running totals, called **accumulators**, pile up. Step through it.

<div class="hz-widget" data-widget="loop-stepper">
<p>▶ Interactive explainer: <a href="https://hizocar.github.io/usm-course-intensivo/classes/Session_1_Day_One/">open this class on the course website</a> to use it.</p>
</div>

```python
total = 0
at_risk = 0

for c in customers:
    total = total + c["monthly_bill"]
    if c["contract"] == "Month-to-month" and c["support_tickets"] >= 3:
        at_risk = at_risk + 1

print(total, at_risk)
print(f"Monthly revenue of the list: {total:.2f} USD. Customers at risk: {at_risk} of {len(customers)}.")
```

```python
# The same loop, one line per customer: the retention team's call sheet
for c in customers:
    flag = "CALL" if c["contract"] == "Month-to-month" and c["support_tickets"] >= 3 else ""
    print(f"{c['customer_id']}  {c['plan']:<10} {c['monthly_bill']:>6.2f} USD  {flag}")
```

Python also has a compact way to build a list from a loop, the **list comprehension**. Read `[c["monthly_bill"] for c in customers]` as *"the bill of every c in customers"*. Add `if ...` at the end to keep only some items. Toggle the explainer above to *comprehension* to watch it happen.

```python
monthly_bills = [c["monthly_bill"] for c in customers]
print(monthly_bills)
print("Total:", sum(monthly_bills))

to_call = [c["customer_id"] for c in customers
           if c["contract"] == "Month-to-month" and c["support_tickets"] >= 3]
print("Who should we call?", to_call)
```

**Business takeaway:** the loop and the comprehension give the same answer (146.35 USD, 2 customers to call). Use whichever you can read more easily; comprehensions shine for short "take this from every row" jobs.

## <span class="hz-step-n">07</span> Functions: write the rule once

We have now typed the retention rule three times. A **function** packages it under a name: you write it once with `def`, and **call** it for any customer. What goes in are the **arguments**, what comes out is the **return value**.

The commercial team needs a second function: a quick **bill estimate** from Cordillera's price list (plan base price plus 0.55 USD per GB of data, only 0.10 USD per GB on Unlimited, plus 8 USD for roaming). A parameter can have a **default** value, used when the caller leaves it out: watch the machine with `roaming` left out, then tick `roaming=True`.

<div class="hz-widget" data-widget="function-machine">
<p>▶ Interactive explainer: <a href="https://hizocar.github.io/usm-course-intensivo/classes/Session_1_Day_One/">open this class on the course website</a> to use it.</p>
</div>

```python
def risk_level(customer):
    """Retention rule: 'at risk', 'watch' or 'ok'."""
    month_to_month = customer["contract"] == "Month-to-month"
    many_tickets = customer["support_tickets"] >= 3
    if month_to_month and many_tickets:
        return "at risk"
    elif month_to_month or many_tickets:
        return "watch"
    else:
        return "ok"


for c in customers:
    print(c["customer_id"], "→", risk_level(c))
```

```python
def estimate_bill(plan, data_gb, roaming=False):
    base = {"Prepaid": 9, "Basic": 18, "Plus": 29, "Unlimited": 42}[plan]
    per_gb = 0.10 if plan == "Unlimited" else 0.55
    bill = base + per_gb * data_gb
    if roaming:
        bill = bill + 8
    return round(bill, 2)


estimate = estimate_bill("Plus", 5.2)
print("Plus, 5.2 GB:", estimate)
print("Plus, 5.2 GB, roaming:", estimate_bill("Plus", 5.2, roaming=True))
print("Unlimited, 30 GB:", estimate_bill("Unlimited", 30))
```

CM-10482 used 5.2 GB last month on Plus, so the price list says about 31.86 USD; the real bill was 31.9. The small gap comes from things the estimate ignores (call minutes, contract discounts). In session 3 you will let the data **learn** these prices on its own, with a regression model: that is the commercial team's question.

Functions combine with comprehensions, too: one line gives the risk level of the whole list.

```python
levels = [risk_level(c) for c in customers]
print(levels)
print("At risk:", levels.count("at risk"), "| watch:", levels.count("watch"), "| ok:", levels.count("ok"))
```

## <span class="hz-step-n">08</span> NumPy: thousands of customers at once

Five customers fit in a loop. Cordillera has **five thousand**, and every month adds five thousand new bills. Python loops handle one item at a time, which gets slow at that scale.

**NumPy** stores numbers in an **array** and runs an operation on all of them at once, in fast compiled code: `bills * 1.05` raises every price in one step, `bills > 40` checks every bill in one step. Race the two approaches.

<div class="hz-widget" data-widget="vectorize">
<p>▶ Interactive explainer: <a href="https://hizocar.github.io/usm-course-intensivo/classes/Session_1_Day_One/">open this class on the course website</a> to use it.</p>
</div>

```python
import numpy as np

bills = np.array([29.4, 30.8, 30.1, 35.6, 32.3, 31.9])   # CM-10482, April to September

print(bills * 1.05)          # every bill after a 5% rise, no loop
print(bills.mean(), bills.max())
print(bills > 31)            # which months were above 31 USD?
```

Now the real scale. We simulate one month for 5,000 customers with Cordillera's price list (step 09 swaps the simulation for the real file). `base_prices[plan_idx]` looks up 5,000 base prices at once, and `np.where` picks the per-GB price for every customer: this is `estimate_bill` without a loop.

```python
rng = np.random.default_rng(42)
n = 5000

plan_idx = rng.choice(4, size=n, p=[0.25, 0.30, 0.28, 0.17])   # 0 Prepaid, 1 Basic, 2 Plus, 3 Unlimited
data_gb = rng.gamma(4.0, 3.0, size=n).round(1)

base_prices = np.array([9, 18, 29, 42])
per_gb = np.where(plan_idx == 3, 0.10, 0.55)
month_bills = (base_prices[plan_idx] + per_gb * data_gb).round(2)

print(month_bills[:8])
print(f"{len(month_bills):,} bills, monthly revenue {month_bills.sum():,.2f} USD, average {month_bills.mean():.2f} USD")
```

Business questions become one-liners. What does a 5% price rise bring in? How many customers pay more than 40 USD? A comparison gives a **boolean mask**, an array of `True`/`False`; `sum` counts the `True`s and `mean` gives their share.

```python
extra = (month_bills * 1.05).sum() - month_bills.sum()
print(f"A 5% rise adds {extra:,.2f} USD a month")

high = month_bills > 40
print(f"{high.sum():,} customers pay more than 40 USD ({high.mean():.1%} of the base)")
print("Their average bill:", round(month_bills[high].mean(), 2))
```
