# Seaborn

## Introduction

[Seaborn](https://seaborn.pydata.org/) is a Python data visualization library built on top of Matplotlib. It provides a high-level interface for drawing attractive and informative statistical graphics, and it is widely used in data science because it works directly with pandas DataFrames.

## Example

Here is a complete example of how to use `seaborn` with the **iris** dataset:

```python
# Import the libraries we need
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_style("whitegrid")
```

```python
# Load the iris dataset that ships with Seaborn
iris = sns.load_dataset("iris")
iris.head()
```

```python
# Bar chart: average sepal length by species
sns.barplot(data=iris, x='species', y='sepal_length')
plt.title('Average Sepal Length by Species')
plt.xlabel('Species')
plt.ylabel('Average sepal length')
plt.show()
```

```python
# Histogram of sepal length
sns.histplot(data=iris, x='sepal_length', kde=True)
plt.title('Sepal Length Histogram')
plt.xlabel('Sepal length')
plt.ylabel('Frequency')
plt.show()
```

```python
# Line chart: petal length across the row index
sns.lineplot(data=iris, x=iris.index, y='petal_length', hue='species')
plt.title('Petal Length Across Rows')
plt.xlabel('Index')
plt.ylabel('Petal length')
plt.show()
```

```python
# Scatter plot: sepal length vs. sepal width
sns.scatterplot(data=iris, x='sepal_length', y='sepal_width', hue='species')
plt.title('Scatter Plot: Sepal Length vs. Sepal Width')
plt.xlabel('Sepal length')
plt.ylabel('Sepal width')
plt.show()
```

```python
# Box plot: distribution of petal length by species
sns.boxplot(data=iris, x='species', y='petal_length')
plt.title('Box Plot: Petal Length by Species')
plt.xlabel('Species')
plt.ylabel('Petal length')
plt.show()
```

```python
# Violin plot: distribution of petal length by species
sns.violinplot(data=iris, x='species', y='petal_length', inner='quart')
plt.title('Violin Plot: Petal Length by Species')
plt.xlabel('Species')
plt.ylabel('Petal length')
plt.show()
```

```python
# One scatter plot per species, side by side
g = sns.FacetGrid(iris, col='species', height=4)
g.map(sns.scatterplot, 'sepal_length', 'sepal_width', alpha=0.7)
g.set_axis_labels('Sepal length', 'Sepal width')
g.set_titles(col_template='{col_name}')
plt.show()
```
